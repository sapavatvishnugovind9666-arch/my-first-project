# Question 1

```python
#import matplotlib library for creating plots
import matplotlib.pyplot as plt
# a BOX PLOT

#create a list of numerical data
data = [11,13,15,17,19,16,18,12,14,21]

# Create a box plot
# widths=0.5       -> Sets the width of the box
# notch=True       -> Displays a notch around the median
# vert=True        -> Creates a vertical box plot
# patch_artist=True -> Fills the box with color
# showmeans=True   -> Displays the mean value
# showfliers=True  -> Displays outliers
plt.boxplot(data,widths=0.5,notch=True,vert=True,patch_artist=True,showmeans=True,showfliers=True)

# Add a title to the box plot
plt.title('BOX PLOT')
#label the y axis
plt.ylabel('Values')
plt.show()


# histogram

# Create a histogram using the same data
# bins=8          -> Divides the data into 8 intervals
# alpha=0.7       -> Sets transparency
# density=False   -> Shows frequency instead of probability density
# cumulative=False -> Creates a normal histogram
# edgecolor='black' -> Adds black borders to the bars
data = [11,13,15,17,19,16,18,12,14,21]
plt.hist(data,bins=8,alpha=0.7,density=False,cumulative=False,edgecolor='black')
plt.title('HISTOGRAM PLOT')
plt.ylabel('Values')
plt.show()
```

# Question 2

```python
#import matplot for plotting graphs
import matplotlib.pyplot as plt
# histogram
#import pandas for reading and working with csv files
import pandas as pd

#Read the mtcars.csvfile into a dataframe
data = pd.read_csv('mtcars.csv')

# Create a histogram of the 'wt' (weight) column
# bins=8 -> Divides the data into 8 intervals
# alpha=0.5 -> Makes the bars slightly transparent
# density=False -> Displays frequency
# cumulative=False -> Creates a normal histogram
# edgecolor='black' -> Adds borders to the bars
plt.hist(data['wt'],bins=8,alpha=0.5,density=False,cumulative=False,edgecolor='black')

#add a title
plt.title('HISTOGRAM PLOT')
#label y axis
plt.ylabel('Values')
plt.show()

# BOX PLOT
# Select MPG values for cars having 4 cylinders
mpg_4=data[data['cyl']==4]['mpg']
# Select MPG values for cars having 6 cylinders
mpg_6=data[data['cyl']==6]['mpg']
# Select MPG values for cars having 8 cylinders
mpg_8=data[data['cyl']==8]['mpg']

# Create box plots for the three cylinder groups
# tick_labels -> Names displayed below each box
# patch_artist=True -> Fills the boxes
# widths=0.5 -> Sets box width
# showmeans=True -> Shows the mean
# showfliers=True -> Shows outliers
plt.boxplot([mpg_4,mpg_6,mpg_8],tick_labels=['4 cylinders','6 cylinders','8 cylinders'],patch_artist=True,widths=0.5,showmeans=True,showfliers=True)
# Add a title
plt.title('BOX PLOT')
# Display the box plot
plt.ylabel('Values')
plt.show()
```

# Question 3

```python
# Import matplotlib for creating graphs
import matplotlib.pyplot as plt

# a BOX PLOT
# Import pandas for handling CSV data
import pandas as pd
# Read the heightweight.csv file
data = pd.read_csv('heightweight.csv')

# Create box plots for height and weight
# The first box represents height
# The second box represents weight
plt.boxplot([data['heightIn'],data['weightLb']],tick_labels=['heights','weights'],showmeans=True)
#add title
plt.title('BOX PLOT')
#label the x axis
plt.xlabel('Measurements')
#label the y axis
plt.ylabel('Values')
plt.show()


# HISTOGRAM

# Create a histogram for height values
# bins=10 -> Divides the data into 10 intervals
# alpha=0.5 -> Sets transparency
# density=False -> Displays frequency
# cumulative=False -> Creates a normal histogram
# edgecolor='black' -> Adds borders to bars
plt.hist([data['heightIn']],bins=10,alpha=0.5,density=False,cumulative=False,edgecolor='black')
#add title
plt.title('HISTOGRAM PLOT')
#label the x axis
plt.xlabel('Measurements')
#label y axis
plt.ylabel('Values')
plt.show()

# Create a histogram for weight values
plt.hist([data['weightLb']],bins=10,alpha=0.5,density=False,cumulative=False,edgecolor='black')
# add title
plt.title('HISTOGRAM PLOT')
#label x axis
plt.xlabel('Measurements')
#label y axis
plt.ylabel('Values')
plt.show()
```

```python
# Import matplotlib for plotting
import matplotlib.pyplot as plt
# a BOX PLOT
# Import pandas for reading CSV files
import pandas as pd
# Read the housing_data.csv file
data = pd.read_csv('housing_data.csv')

# Set the size of the figure
# figsize=(20,10) means width=20 and height=10
plt.figure(figsize=(20,10))

# Create box plots of SalePrice for each Neighborhood
# column='SalePrice' -> The numerical data being plotted
# by='Neighborhood' -> Groups the data according to neighborhood
data.boxplot(column='SalePrice',by='Neighborhood')
#add title
plt.title('BOX PLOT')
#label x axis
plt.ylabel('Saleprice')

# Rotate neighborhood names by 90 degrees
# This makes long names easier to read
plt.xticks(rotation=90)
plt.show()
```

```python
# pie chart
# Import matplotlib for creating the pie chart
import matplotlib.pyplot as plt
# Import pandas for handling CSV data
import pandas as pd
# Read the housing_data.csv file
data = pd.read_csv('housing_data.csv')

# Count the number of houses in each MSZoning category
# value_counts() returns the frequency of each category
zoning_columns=data['MSZoning'].value_counts()

# Create a pie chart
# zoning_columns -> Values used to determine slice sizes
# labels -> Names of the zoning categories
# autopct='%1.1f%%' -> Displays percentage with one decimal place
plt.pie(zoning_columns,labels=zoning_columns.index,autopct='%1.1f%%')
plt.show()
```
