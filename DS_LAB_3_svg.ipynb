```python
#weekly sales data

#store sales values in a list
sales_data=[120,150,130,170,160,180,140]

#store the days of a week
days_of_week=[
    'monday','tuesday','wednesday','thursday','friday','saturday','sunday'
]
# (a) Create pandas series using the sales data
sales=pd.Series(sales_data)

#  Assign days of a week as the index
sales.index =days_of_week

print("weekly sales:",sales)
```

```python
#(b) Access the sales for monday and saturday
print(sales['monday'])
print(sales['saturday'])

#Calculate total sales
total_sales=sales.sum()
print(total_sales)

#find the heighest and lowest sales
hieghest_sales=sales.max()
lowest_sales=sales.min()

# Find the days with heighest and lowest sales
hieghest_day=sales.idxmax()
lowest_day=sales.idxmin()


print(hieghest_sales)
print(lowest_sales)
print(hieghest_day)
print(lowest_day)
```

```python
# (c) Calculate the average values
average_sale=sales.mean()
print(average_sale)

#Calculate the difference between each day's sales and average sales
difference=abs(sales-average_sale)
print(difference)

#Display the days where the difference is greater than than 20
print(sales[difference > 20])
```

```python
# Import the pandas library
import pandas as pd

# Read the AmesHousing CSV file into a DataFrame
df=pd.read_csv("AmesHousing.csv")

# Display the first 5 rows of the dataset
display(df.head())

# Display the number of rows and columns
print(df.shape)

# Display all column names
print(df.columns)

# Display the data type of each column
print(df.dtypes)

# Display statistical summary of numerical columns
print(df.describe())
```

```python
# c
# Display all column names with their exact representation
# This helps identify unwanted spaces in column names
print([repr(col) for col in
       df.columns])

# Remove leading and trailing spaces from column names
df.columns = df.columns.str.strip()

# Select the required numerical columns
columns=["Lot Area","Year Built","1st Flr SF","2nd Flr SF","SalePrice"]

# Calculate and display the mean of the selected columns
print("mean:")
print(df[columns].mean())

# Calculate and display the median of the selected columns
print("median:")
print(df[columns].median())

# Calculate and display the mode of the selected columns
print("mode:")
print(df[columns].mode())

# Calculate and display the standard deviation
print("standard deviation:")
print(df[columns].std())

# Calculate and display the variance
print("variance:")
print(df[columns].var())
```

```python
#d

# Select the columns required for correlation analysis
columns=["Lot Area","Year Built","1st Flr SF","2nd Flr SF","SalePrice"]

# Calculate the correlation between the selected columns
# Display only the correlation values with SalePrice
print(df[columns].corr()["SalePrice"])
```

```python
#e
# Display the frequency of houses in each neighborhood
print("Neighborhood:")
print(df["Neighborhood"].value_counts())

# Display the frequency of each building type
print(df["Bldg Type"].value_counts())

# Display the frequency of each house style
print(df["House Style"].value_counts())
```
