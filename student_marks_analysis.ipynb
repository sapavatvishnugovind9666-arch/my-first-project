```python
marks=[85,90,95,87,94]
average= sum(marks)/len(marks)
print("marks:",marks)
print("average:",average)
print("heighest:",max(marks))
print("lowest:",min(marks))
```

```python
import matplotlib.pyplot as plt
students=["student1","student2","student3","student4","student5"]
marks=[85,90,95,87,94]
plt.bar(students,marks)
plt.xlabel("students")
plt.ylabel("marks")
plt.title("marks of students")
plt.show()
```

```python
import pandas as pd
data = {
    "Student": ["A", "B", "C", "D", "E"],
    "Maths": [85, 72, 90, 65, 78],
    "Science": [88, 75, 92, 70, 80],
    "English": [80, 82, 85, 68, 75]
}

df = pd.DataFrame(data)

print("Highest marks in each subject:")
print(df[["Maths", "Science", "English"]].max())

print("\nLowest marks in each subject:")
print(df[["Maths", "Science", "English"]].min())
```

```python
import matplotlib.pyplot as plt
df.set_index("Student")[["Maths","Science","English"]].plot(kind='bar',figsize=(9,5))
plt.title("student marks")
plt.xlabel("student")
plt.ylabel("marks")
plt.show()
```

```python
import pandas as pd
data = {
    "Student": ["A", "B", "C", "D", "E"],
    "Maths": [85, 72, 90, 65, 78],
    "Science": [88, 75, 92, 70, 80],
    "English": [80, 82, 85, 68, 75]
}

df = pd.DataFrame(data)
df['Average']=df[["Maths","Science","English"]].mean(axis=1)
print(df)
```
