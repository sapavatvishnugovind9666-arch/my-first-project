# QUESTION 1

```python
# Import matplotlib for creating plots
import matplotlib.pyplot as plt
# (A)
# Define the X-axis values
x=[1,2,3,4,5]
# Define the Y-axis values
y=[10,20,32,12,23]
# Create a line plot with red color, circle markers, dashed line, and linewidth 2
plt.plot(x,y,color='r',marker='o',linewidth=2,linestyle='--')
# Add a label to the X-axis
plt.xlabel('X Axis')
# Add a label to the Y-axis
plt.ylabel('Y Axis')
# Add a title to the graph
plt.title('Line Plot')
plt.show()
```

```python
# (B)
# Store X-axis values
x=[1,2,3,4,5]
# Store Y-axis values
y=[12,23,31,20,27]
# Create a scatter plot
plt.scatter(x,y,color='g',marker='o')
# Set X-axis label
plt.xlabel('X Axis')
# Set Y-axis label
plt.ylabel('Y Axis')
# Set the title of the graph
plt.title('scatter plot')
plt.show()
```

```python
# (C)
# Store student names
name=['A','B','C','D','E']
# Store marks of students
marks=[76,50,96,85,65]
# Create a bar chart
plt.bar(name,marks,color='y')
# Set X-axis label
plt.xlabel('name')
# Set Y-axis label
plt.ylabel('marks')
# Set the title of the graph
plt.title('student report')
plt.show()
```

# QUESTION 3

```python
# Read the mtcars CSV file
df=pd.read_csv('mtcars.csv')
# Display the column names
df.columns
```

```python
# (A)
 # Create a scatter plot of HP and MPG
plt.scatter(df['hp'],df['mpg'],color='r',s=100,marker='o')
# Set the X-axis label
plt.xlabel('horse power')
# Set the Y-axis label
plt.ylabel('MPG')
 # Set the graph title
plt.title('HP PER MPG')
plt.show()
```

```python
# (B)
# Count cars for each number of gears
gear_count=(df['gear'].value_counts().sort_index())
# Create a bar chart
plt.bar(gear_count.index.astype(str),gear_count.values)
# Set the X-axis label
plt.xlabel('no of gears')
# Set the Y-axis label
plt.ylabel('np of cars')
# Set the graph title
plt.title('car spesifications by no of gears')
plt.show()
```

# QUESTION 5

```python
# Read the housing dataset
df=pd.read_csv('housing_data.csv')
 # Display all column names
df.columns
```

```python
#(A)
 # Calculate the average sale price for each year
means=df.groupby('YrSold')['SalePrice'].mean()
# Plot the average house prices against the year
plt.plot(means.index,means.values)
# Set the X-axis label
plt.xlabel('YEAR')
# Set the Y-axis label
plt.ylabel('price')
# Set the graph title
plt.title('TREND OF HOUSE PRICES OVER THE YEARS')
plt.show()
```

```python
#(B)
# Create a scatter chart using the LotArea and SalePrice
plt.scatter(df['LotArea'],df['SalePrice'],color='g',alpha=0.3)
# Set the X-axis label
plt.xlabel('LotArea')
# Set the Y-axis label
plt.ylabel('price')
# Set the title of the graph
plt.title('Relation btw LotArea and SalePrice')
plt.show()
```

```python
# (C)
# Count the number of houses in each neighborhood
Neighborhood_count=df['Neighborhood'].value_counts()
# Create a bar chart for the number of houses in each neighborhood
plt.bar(Neighborhood_count.index,Neighborhood_count.values)
# Set the X-axis label
plt.xlabel('Neighborhood')
# Set the X-axis label
plt.ylabel('no of houses')
# Set the title of the graph
plt.title('No of houses in each Neighborhood')
# Rotate the X-axis labels for better readability
plt.xticks(rotation=90)
plt.show()
```
